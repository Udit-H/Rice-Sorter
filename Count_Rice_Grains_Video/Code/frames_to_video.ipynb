```python
import cv2
import os
```

```python
def frames_to_video(frames_folder, output_video_path, fps=20):
    # Get all image files in order
    images = sorted([
        img for img in os.listdir(frames_folder)
        if img.endswith((".png", ".jpg", ".jpeg"))
    ])

    if len(images) == 0:
        print("No images found!")
        return

    # Read first image to get size
    first_frame = cv2.imread(os.path.join(frames_folder, images[0]))
    height, width, _ = first_frame.shape

    # Define video writer
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    video = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

    for image_name in images:
        image_path = os.path.join(frames_folder, image_name)
        frame = cv2.imread(image_path)

        if frame is None:
            continue

        video.write(frame)

    video.release()
    print("Video saved at:", output_video_path)
```

```python
frames_folder = "original_output_frames/detected"  # your processed frames
output_video = "original_output_video.mp4"

frames_to_video(frames_folder, output_video, fps=60)
```
